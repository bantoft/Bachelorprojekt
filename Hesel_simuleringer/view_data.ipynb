```python
import xbout
import matplotlib.pyplot as plt

ds = xbout.open_boutdataset('BOUT-HESEL/data/BOUT.dmp.*.nc')
```

```python
n  = (ds['lnpi'].isel(y=0, t=-1).values).T
pe = (ds['lnpe'].isel(y=0, t=-1).values).T

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(n,  origin='lower', aspect='auto'); axes[0].set_title('lnpi (t=last)')
axes[1].imshow(pe, origin='lower', aspect='auto'); axes[1].set_title('lnpe (t=last)')
plt.tight_layout()
plt.show()
```
